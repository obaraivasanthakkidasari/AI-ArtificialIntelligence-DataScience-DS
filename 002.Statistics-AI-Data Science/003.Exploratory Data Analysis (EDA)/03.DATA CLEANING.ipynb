```python
df=pd.read_csv("claimants.csv")
df
```

```python
from sklearn.impute import SimpleImputer
```

```python
mean_imputer=SimpleImputer(strategy='mean')

df["AGE"]=mean_imputer.fit_transform(df[["AGE"]])
df
```

```python
median_imputer=SimpleImputer(strategy='median')

df["LOSS"]=median_imputer.fit_transform(df[["LOSS"]])
df
```

```python
mode_imputer=SimpleImputer(strategy='most_frequent')

df["INSUR"]=mode_imputer.fit_transform(df[["INSUR"]])
df
```

# OUTLIERS

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("claimants.csv")
df
```

```python
df.isnull().sum()
```

```python
#Various ways o find outliers
#BOX PLOT
```

```python
sns.boxplot(x=df["LOSS"])
plt.show()
```

```python
    Q1=df["LOSS"].quantile(0.25)
    Q3=df["LOSS"].quantile(0.75)
    IQR=Q3-Q1
    print(Q1,Q3,IQR)
```

```python
lower_limit=Q1-(IQR*1.5)
upper_limit=Q3+(IQR*1.5)
lower_limit,upper_limit
```

```python
df[df["LOSS"]>upper_limit]
```

```python
df[df["LOSS"]<lower_limit]
```

```python
df['LOSS'].fillna(df["LOSS"].median(),inplace=True)
df
```

```python
sns.boxplot(x=df["LOSS"])
plt.show()
#Box plot is never change that is the meaning of above syntax
```

# Dealing with Outliers

```python
#1.REMOVE
```

```python
df_trimmed=df[(df["LOSS"]>lower_limit)&(df["LOSS"]<upper_limit)]
df_trimmed
```

```python
sns.boxplot(x=df_trimmed["LOSS"])
plt.show()
#Recalculated
```

```python
win= Winsorizer(capping_method='iqr',tail='both',fold=1.5)

df["LOSS"]=win.fit_transform(df[["LOSS"]])

df
```
