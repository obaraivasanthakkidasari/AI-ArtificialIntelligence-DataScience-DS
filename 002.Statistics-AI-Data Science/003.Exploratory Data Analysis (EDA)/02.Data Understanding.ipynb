4. Identify Variable Types
The PDF’s final line contains a typo (tips instead of tip and mismatched quotes). 
Below is corrected executable code. Continuous: total_bill, tip. Discrete numerical: size (party size). 
Categorical: sex, smoker, day, time. 
The source grouped size with the other noncontinuous columns; here it is separated because it is a count.

```python
continuous = ["total_bill", "tip"]
discrete = ["size"]
categorical = ["sex", "smoker", "day", "time"]

print("Continuous:", continuous)
print("Discrete:", discrete)
print("Categorical:", categorical)
```
