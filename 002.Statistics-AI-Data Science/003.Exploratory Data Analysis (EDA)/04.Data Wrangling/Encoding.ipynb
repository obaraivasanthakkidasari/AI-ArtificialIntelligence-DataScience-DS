# Nominal Encoding

```python
#Convert nominal categorical into numerical

#Option1: Get_dummies() using pandas
```

```python
df_dum=pd.get_dummies(df,drop_first=True)
df_dum
```

```python
#EXPLANATION Part get_dummies()
```

```python
dummies=pd.get_dummies(df.town)
dummies
```

```python
df_dummies=pd.concat([df,dummies],axis='columns')
df_dummies
```

```python
df_dummies.drop('town',axis='columns',inplace=True)
df_dummies
```

```python
df_dummies.drop('Banglore',axis='columns',inplace=True)
df_dummies
```

```python
#Option 2: OneHotEncoder using Sklearn
```

```python
from sklearn.preprocessing import OneHotEncoder

enc=OneHotEncoder(drop='first')
```

```python
enc_df=pd.DataFrame(enc.fit_transform(df[['town']]).toarray())
enc_df
```

```python
#Merge with main df
df_ohe=pd.concat([df,enc_df],axis='columns')
df_ohe
```

```python
#Drop the original variable
df_ohe.drop('town',axis='columns',inplace=True)
df_ohe
```

# 2.ORDINAL to NUMERIC Encoding

```python
#Label encoder using sklearn
df=pd.read_csv('homeprices.csv')

##Import from sklearn library
from sklearn.preprocessing import LabelEncoder

##call the function
le=LabelEncoder()

##fit_transform
df["town"]=le.fit_transform(df["town"])
df
```

```python
#Not alphabetical order
```

```python
#Ordinal encoder using sklearn
df=pd.read_csv('homeprices.csv')

##Import from sklearn library
from sklearn.preprocessing import OrdinalEncoder

##call the function
oe=OrdinalEncoder(categories=[['Chennai','Hyderabad','Banglore']])

##fit_transform
df["town"]=oe.fit_transform(df[["town"]])
df
```

# OPTION 3.PD.MAP

```python
df=pd.read_csv('homeprices.csv')

df['town']=df['town'].map({'Banglore':2,'Hyderabad':1,'Chennai':0})
df
```
