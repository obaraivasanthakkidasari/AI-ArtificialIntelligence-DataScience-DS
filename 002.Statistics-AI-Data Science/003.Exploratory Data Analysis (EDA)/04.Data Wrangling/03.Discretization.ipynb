# Creating Bins

```python
intervals=[0,12,19,30,60,90]
categories=['child','teenager','young_adult','middle_aged','senior_citizen']
```

# Discretization

```python
df['age_category']=pd.cut(df['age'],bins=intervals,labels=categories)
```
