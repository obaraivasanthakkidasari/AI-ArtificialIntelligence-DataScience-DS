# FEATURE TRANSFORMATION

```python
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv('TITANICDW.csv')
data.head()
```

```python
data=pd.read_csv('TITANICDW.csv',usecols=['Fare'])
data.head()
```

```python
data['Fare'].skew()
```

# Log Transformation

```python
data['Log_Fare']=np.log(data['Fare']+0.0001)
data['Log_Fare'].skew()
```

# Root Transformation

```python
data['sqr_Fare']=data['Fare']**(1/4.7)
data['sqr_Fare'].skew()
```

# Reciprocal Transformation

```python
data['Rec_Fare']=1/(data['Fare']+0.01)
data['Rec_Fare'].skew()
```

# BoxCox

```python
from scipy.stats import boxcox
```

```python
#Transform training data & save lambda value
data['Fare_boxcox'], param=boxcox(data.Fare+0.01)
```

```python
print("lambda =",param)
```

```python
data['Fare_boxcox'].skew()
```
